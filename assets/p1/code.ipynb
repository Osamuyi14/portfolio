#Data Preparation

##Data Reading
I did data reading to load and understand my data.

```python
from pandas import read_csv, get_dummies, DataFrame
student = read_csv('/content/drive/MyDrive/StudentPerformanceFactors.csv')
student
student.head()
student.shape
student.info()
```

##Feature Selection
This is done to remove the irrelevent data, and in this case the only irrelevant data was "Gender, School type, Physical activity, Sleep hours, Distance from home"

```python
student = student.drop({"Gender", "School_Type", "Physical_Activity", "Sleep_Hours", "Distance_from_Home"}, axis=1)
student.info()
```

##Data Cleaning
I am doing this to clean the data by filling in all the missing elements from the dataset. Teacher_Quality and Parental_Education_level are the only columns with missing data.

```python
missing1 = student['Teacher_Quality'].mode()
student['Teacher_Quality'] = student['Teacher_Quality'].fillna(missing1[0])
student.info()
```

```python
missing2 = student['Parental_Education_Level'].mode()
student['Parental_Education_Level'] = student['Parental_Education_Level'].fillna(missing2[0])
student.info()
```

##Data Encoding
This is done because models don't read  read/understand letter, so it has to be converted to number before the models can computate it.

```python
student['Parental_Involvement'] = student['Parental_Involvement'].map({"Low":0, "Medium":1, "High":2})
student.info()
```

```python
student['Access_to_Resources'] = student['Access_to_Resources'].map({"Low":0, "Medium":1, "High":2})
student.info()
```

```python
student['Extracurricular_Activities'] = student['Extracurricular_Activities'].map({"No":0, "Yes":1})
student.info()
```

```python
student['Motivation_Level'] = student['Motivation_Level'].map({"Low":0, "Medium":1, "High":2})
student.info()
```

```python
student['Internet_Access'] = student['Internet_Access'].map({"No":0, "Yes":1})
student.info()
```

```python
student['Family_Income'] = student['Family_Income'].map({"Low":0, "Medium":1, "High":2})
student.info()
```

```python
student['Teacher_Quality'] = student['Teacher_Quality'].map({"Low":0, "Medium":1, "High":2})
student.info()
```

```python
student['Peer_Influence'] = student['Peer_Influence'].map({"Neutral":0, "Negative":1, "Positive":2})
student.info()
```

```python
student['Learning_Disabilities'] = student['Learning_Disabilities'].map({"No":0, "Yes":1})
student.info()
```

```python
student['Parental_Education_Level'] = student['Parental_Education_Level'].map({"High School":0, "College":1, "Postgraduate":2})
student.info()
```

##Divide data into x and y
I did this so my input and output elements will be known

```python
y = student["Exam_Score"]
x = student.drop('Exam_Score', axis=1)
print(x.shape)
print(y.shape)
```

##Data Scaling (only x)
This was done to normalise the data

```python
from sklearn.preprocessing import StandardScaler
x_scaled = StandardScaler().fit_transform(x)
DataFrame(x_scaled)
```

##Data Spliting
I did this to divide my dataset into the training and test sets.

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train,y_test = train_test_split(x_scaled ,y,test_size =0.10, random_state=100)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

There is no data balancing in regression

#Decision Tree Regressor

##Method 1

###Modelling

```python
from sklearn.tree import DecisionTreeRegressor
DT_Regressor1 = DecisionTreeRegressor(max_depth=3) # building set for the decision tree
DT_Regressor1.fit(x_train,y_train) # training set for the decision tree
y_pred1 = DT_Regressor1.predict(x_test)   #testing set for the decision tree
print(y_pred1)
```

##Evaluation

```python
from sklearn.metrics import r2_score
R2 = r2_score(y_test,y_pred1)
print('R2 = ',round(R2*100,2),'%')
```

##Method 2

```python
from sklearn.model_selection import GridSearchCV
DT_Regressor2 = DecisionTreeRegressor()
depths = {'max_depth':range(1,100)}

grid_search1 =GridSearchCV(estimator=DT_Regressor2, param_grid=depths,scoring ='r2', cv=5 )      #building: 4 components
grid_search1.fit(x_scaled,y) #fitting : 4 tasks : training, testing , evaluation and , ranking for the decision tree

best_depth = grid_search1.best_params_
print(best_depth)
R2 = grid_search1.best_score_
print('R2 = ',round(R2*100,2),'%')
best_tree = grid_search1.best_estimator_
print(best_tree)
```
